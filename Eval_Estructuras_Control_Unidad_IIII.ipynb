# Ejercicio 2. Registro y análisis de números

## Enunciado

Crea un programa que solicite cuántos números desea analizar el usuario. La cantidad debe ser un entero mayor que cero. Si no lo es, muestra un mensaje de error y termina el análisis.

Si la cantidad es válida, utiliza un ciclo `for` para solicitar esa cantidad de números. A medida que se ingresan, el programa debe:

1. Contar cuántos son positivos, cuántos son negativos y cuántos son iguales a cero.
2. Calcular la suma de todos los números.
3. Calcular el promedio de los números ingresados.

Utiliza una estructura condicional dentro del ciclo para clasificar cada número. El promedio se calcula dividiendo la suma entre la cantidad de números ingresados.

## Ejemplo de entrada

```text
¿Cuántos números desea analizar?: 5
Ingrese el número 1: 4
Ingrese el número 2: -2
Ingrese el número 3: 0
Ingrese el número 4: 7
Ingrese el número 5: 1
```

## Ejemplo de salida

```text
Cantidad de positivos: 3
Cantidad de negativos: 1
Cantidad de ceros: 1
Suma total: 10
Promedio: 2.00
```

## Consideraciones

- Inicializa los contadores y el acumulador antes de iniciar el ciclo.
- No guardes los números en una lista: practica el uso de contadores y acumuladores.
- Si la cantidad indicada es cero o negativa, no solicites los números que se van a analizar.

````python
cantidad = int(input("Por favor, ingrese la cantidad de numeros a analizar: "))

if cantidad > 0:
    numeros = 0
    positivos = 0
    negativos = 0
    ceros = 0

    for i in range (cantidad):
        valor = int(input(f"ingrese el numero {i + 1}: "))
        numeros += valor

        if valor > 0:
            positivos += 1
        elif valor < 0:
            negativos += 1
        elif valor == 0:
            ceros += 1

    promedio = numeros / cantidad       
    print (f"Cantidad de positivos: {positivos}")
    print (f"Cantidad de negativos: {negativos}")
    print (f"Cantidad de ceros: {ceros}")
    print (f"Suma total: {numeros}")
    print (f"Promedio: {promedio}")

  
else:
    print("¡ERROR! VALOR INVALIDO.")  
````

# Ejercicio 4. Analizador de contraseñas

## Enunciado

Desarrolla un programa que solicite una contraseña y evalúe sus características. Debes recorrer la contraseña carácter por carácter y utilizar métodos de cadena de Python para identificar los tipos de caracteres.

El programa debe comprobar estos cuatro criterios:

1. La contraseña tiene **ocho caracteres o más**.
2. Contiene al menos una letra mayúscula.
3. Contiene al menos una letra minúscula.
4. Contiene al menos un dígito numérico.

Además, debe detectar si la contraseña contiene espacios. Una contraseña con espacios siempre se clasificará como **débil**, aunque cumpla los cuatro criterios anteriores.

Al finalizar, muestra la longitud de la contraseña, si contiene mayúsculas, minúsculas, números y espacios, el número de criterios cumplidos (de 4) y su clasificación según estas reglas:

- **Fuerte:** cumple los cuatro criterios y no contiene espacios.
- **Aceptable:** cumple exactamente tres criterios y no contiene espacios.
- **Débil:** cumple dos criterios o menos, o contiene espacios.

Utiliza un ciclo `for` para recorrer los caracteres y métodos como `.isupper()`, `.islower()`, `.isdigit()` e `.isspace()`. Puedes apoyarte en variables booleanas (`True` o `False`) para registrar lo que encuentres. Utiliza condicionales anidadas para decidir la clasificación final.

## Ejemplo de entrada

```text
Ingrese una contraseña: Clave2026
```

## Ejemplo de salida

```text
Longitud: 9
Contiene mayúscula: Sí
Contiene minúscula: Sí
Contiene número: Sí
Contiene espacios: No
Criterios cumplidos: 4 de 4
Clasificación: Fuerte
```

## Pruebas que debes realizar

Prueba una contraseña corta, otra sin números, otra que no tenga mayúsculas y otra que incluya un espacio. Verifica que la clasificación cambie de acuerdo con las reglas del enunciado.

````python
contrasena = str(input("ingrese una contraseña: "))

mayusculas = False
minusculas = False
numeros = False
espacios = False

for caracter in contrasena:
    if caracter.isupper():
        mayusculas = True 
    elif caracter.islower():
        minusculas = True
    elif caracter.isdigit():
        numeros = True
    elif caracter.isspace():
        espacios = True

longitud = len(contrasena) 

criterios = 0

if longitud >= 8:
    criterios += 1
if mayusculas == True:
    criterios += 1
if minusculas == True:
    criterios += 1
if numeros == True:
    criterios += 1



if espacios == True:
    clasificacion = "Debil"
else:
    if criterios == 4:
        clasificacion = "Fuerte"
    elif criterios == 3:
        clasificacion = "Aceptable"
    else:
        clasificacion = "Debil"        


print (f"Longitud: {longitud}")
print (f"Contiene mayusculas: {mayusculas}")
print (f"Contiene minusculas: {minusculas}")
print (f"Contiene numeros: {numeros}")
print (f"Contiene espacios: {espacios}")
print (f"Criterios cumplidos: {criterios}")
print (f"Clasificacion: {clasificacion}")
````
